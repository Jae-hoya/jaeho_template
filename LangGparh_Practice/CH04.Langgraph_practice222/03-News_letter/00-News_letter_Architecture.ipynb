# 뉴스레터 에이전트 아키텍쳐
- 사용자입력 -> 뉴스 검색 주제설정 -> 세부주제(1~5), 테마설정 에이전트 -> 뉴스검색 -> 본문(1~5) -> 뉴스레터 편집 (본문작성 에이전트)

뉴스테러를 생성하는 에이전트럴 만들기 위해서는 많은 양의 정보를 검색하고, 이를 일련의 연관성있는 소주제로 묶어야한다.
또한, 전체 뉴스레터가 일관된 형식을 갖추고 있으며, 최종적으로 올바른 검토가 필요하다. 이를 위해 여러 에이전트를 묶어 구성한다.

1. 세부주제(search_sub_themes)검색과 이를 기반으로 본문작성은 여러여러번 동작하기 때문에 시간이 오래 걸린다. 따라서, 비동기실행한다.
2. 뉴스레터 에이전트 최종 결과물은 대주제, 소주제로 나누고 그안에 본문을 넣는 형태다.
3. 제일 중요한것은은 뉴스레터의 quality를 높히는것이다. 각 step에 있는 node들에서 세부적인 system_prompt를 정교하게 작성하는것이 매우 중요하다. 
4. 본문작성에도 뉴스레터임을 자각하게 만들어주는 것이 매우 중요하다. 
5. 마지막에 뉴스레터 편집 과정에서 높은 성능을 가진 LLM이 필요할 것이다. (글쓰기는 엔트로픽이 잘 하는것 같다.)
