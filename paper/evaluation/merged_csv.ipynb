```python
import pandas as pd
import os

# 파일 이름 패턴 설정
base_name = "6_Law_Codex"
file_extension = ".csv"
file_range = range(23)  # 0 ~ 22

# 파일을 읽어서 리스트에 저장
dfs = []
for i in file_range:
    filename = f"{base_name}{i}{file_extension}"
    if os.path.exists(filename):  # 파일 존재 확인
        df = pd.read_csv(filename)
        dfs.append(df)
    else:
        print(f"❌ 파일 없음: {filename}")

# 모든 DataFrame을 하나로 합치기
merged_df = pd.concat(dfs, ignore_index=True)

# 결과 저장 (선택사항)
merged_df.to_csv("6_Law_Codex_merged.csv", index=False)

print("✅ 모든 CSV 파일이 병합되었습니다.")
```
