```python
# 오류 설명:
# 현재 PaddleOCR-VL 파이프라인의 predict 함수에 리스트나 여러 파일을 입력하면 오류가 발생함.
# 이유는 해당 파이프라인이 batch_size=1만 지원하기 때문임.
# 즉, 한 번에 하나의 파일(문자열 경로 또는 단일 Path 객체)만 처리할 수 있음.
# 만약 여러 파일을 넣으면 아래와 비슷한 에러가 발생할 수 있음:
# RuntimeError: Exception from the 'input' worker: Only batch_size=1 is supported for inference.
#
# 해결 방법: 파일을 여러 개 처리하려면 for 문 등으로 하나씩 predict()에 전달해야 함.

# 참고: PaddleOCRVLForConditionalGeneration의 모든 가중치는 체크포인트(C:\Users\skyop\.paddlex\official_models\PaddleOCR-VL-1.5)에서 초기화되었습니다.
# 해당 체크포인트 모델이 학습된 작업과 유사하다면, 추가 트레이닝 없이 바로 inference에 사용할 수 있습니다.
# 모델 batch size는 1로만 지원됩니다. (현재 'PaddleOCR-VL-1.5-0.9B' 로컬 모델 기준)
# 설정 파일 로딩: C:\Users\skyop\.paddlex\official_models\PaddleOCR-VL-1.5\generation_config.json

# 단일 파일 또는 여러 파일 경로를 모두 지원하도록 구성
input_files = [
    "_mock_data/sample_document.pdf",
    # "_mock_data/another.pdf",
]
output_path = Path("./output")
output_path.mkdir(parents=True, exist_ok=True)

# GPU/드라이버 이슈를 피하려면 CPU로 고정 (필요 시 device="gpu"로 변경)
# batch_size=1로 고정된 파이프라인 설정을 사용
pipeline = PaddleOCRVL(
    #device="cpu",
    paddlex_config="paddleocr_vl_1_5_bs1.yaml",
)

for file_path in input_files:
    file_path = Path(file_path)
    if not file_path.exists():
        print(f"[SKIP] 파일 없음: {file_path}")
        continue

    # predict() 호출 시 반드시 단일 파일(str)만 넘겨야 함 (batch size=1)
    output = pipeline.predict(str(file_path), batch_size=1)

    for res in output:
        res.print()
        res.save_to_json(save_path=output_path)
        res.save_to_markdown(save_path=output_path)
```

```python
from paddleocr import PaddleOCRVL
from pathlib import Path

input_files = [
    "_mock_data/sample_document.pdf",
]
# 출력 폴더 설정
output_path = Path("./output")
output_path.mkdir(parents=True, exist_ok=True)

# GPU/드라이버 이슈를 피하려면 CPU로 고정 (필요 시 device="gpu"로 변경)
# batch_size=1로 고정된 파이프라인 설정을 사용
pipeline = PaddleOCRVL(
    device="cpu",
    paddlex_config="paddleocr_vl_1_5_bs1.yaml",
)

for file_path in input_files:
    file_path = Path(file_path)
    if not file_path.exists():
        print(f"[SKIP] 파일 없음: {file_path}")
        continue

    output = pipeline.predict(str(file_path), batch_size=1)
    # 결과 저장 및 출력
    for res in output:
        res.print()  # 콘솔에 결과 출력
        res.save_to_json(save_path=output_path)  # JSON으로 저장
        res.save_to_markdown(save_path=output_path)  # Markdown으로 저장
```
