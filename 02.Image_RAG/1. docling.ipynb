# Custom Pipe Line
- OCR은 EasyOCR로 구동됩니다.
- Table 구조 인식은 Docling에서 개발한 Tableformer 모델을 통해 구동됩니다.

```python
import logging
import json
import time
from pathlib import Path

_log = logging.getLogger(__name__)
```

```python
from langchain_core.documents import Document
from docling.datamodel.base_models import InputFormat # 입력 형식 정의에 사용
from docling.datamodel.pipeline_options import PdfPipelineOptions # PDF 처리에 사용
from docling.document_converter import DocumentConverter, PdfFormatOption # 문서변환기 및 pdf 형식 옵션 가져오기

# pdf 파이프라인 옵션 설정
pipeline_options = PdfPipelineOptions()

pipeline_options.do_ocr = False # OCR 끄기. PDF가 드래그가 안되는 스캔된 문서에서 사용. 드래그가 되는 디지털 데이터에 사용 X
pipeline_options.ocr_options.lang = ["ko"] # OCR 사용시 한국어로 설정

pipeline_options.do_table_structure = True # 표 구조 인식 활성화
pipeline_options.table_structure_options.do_cell_matching = True # 표 셀 매칭 활성화

doc_converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)
    }
)
```

```python
start_time = time.time()

input_doc_path = "data/국가별 공공부문 AI 도입 및 활용 전략 (1).pdf"
# 문서 변환 시작
conv_result = doc_converter.convert(input_doc_path)

end_time = time.time() - start_time
# 변환 완료 시간 로깅
_log.info(f"Document converted in {end_time:.2f} seconds.")

## 결과 내보내기
# 출력 디렉토리 설정
output_dir = Path("scratch")
# 출력 디렉토리가 없으면 생성
output_dir.mkdir(parents=True, exist_ok=True)
# 파일 이름 추출 (확장자 제외)
doc_filename = conv_result.input.file.stem

# 문서 JSON 형식으로 내보내기:
with (output_dir / f"{doc_filename}.json").open("w", encoding="utf-8") as fp:
    fp.write(json.dumps(conv_result.document.export_to_dict()))

# 텍스트 형식으로 내보내기:
with (output_dir / f"{doc_filename}.txt").open("w", encoding="utf-8") as fp:
    fp.write(conv_result.document.export_to_text())

# 마크다운 형식으로 내보내기:
with (output_dir / f"{doc_filename}.md").open("w", encoding="utf-8") as fp:
    fp.write(conv_result.document.export_to_markdown())

# doctag 형식으로 내보내기:
with (output_dir / f"{doc_filename}.doctags").open("w", encoding="utf-8") as fp:
    fp.write(conv_result.document.export_to_doctags())
```

```python
# 스캔 문서 OCR해보기 (디지털 데이터)
```

```python
def convert_with_ocr_option(pdf_path, enable_ocr):
    opts = PdfPipelineOptions()
    opts.do_ocr = enable_ocr
    opts.ocr_options.lang = ["ko"]
    opts.do_table_structure = True  # 표 관련 영향 없게 끔
    opts.table_structure_options.do_cell_matching = True
    
    converter = DocumentConverter(
        format_options={InputFormat.PDF: PdfFormatOption(pipeline_options=opts)}
    )
    return converter.convert(pdf_path)

# OCR 꺼짐
res_without_ocr = convert_with_ocr_option("data/은행 확인서 (1).pdf", enable_ocr=False)
print(f"❌ OCR disabled: text blocks = {len(res_without_ocr.document.texts)}")

# OCR 켜짐
res_with_ocr = convert_with_ocr_option("data/은행 확인서 (1).pdf", enable_ocr=True)
print(f"✅ OCR enabled: text blocks = {len(res_with_ocr.document.texts)}")
```

# Docling으로 이미지까지 처리하기
Docling의 picture_description_options 옵션을 활용하면, 페이지에서 추출된 이미지에 대한 캡션과 주석을 생성할 수 있다.

OpenAI나 Claude와 같이 Vision을 지원하는 API를 활용할 수 있다.

```python
from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import (
    PdfPipelineOptions,
    PictureDescriptionApiOptions,
)
from docling.document_converter import DocumentConverter, PdfFormatOption
import os

# 1. OpenAI Vision caption 옵션 설정
openai_api_key = os.getenv("OPENAI_API_KEY")
openai_vlm_options = PictureDescriptionApiOptions(
    url="https://api.openai.com/v1/chat/completions",
    params=dict(
        model="gpt-4.1",
    ),
    headers={"Authorization": f"Bearer {openai_api_key}"},  # 실제 키 필요
    prompt="Describe this image in two sentences. Focus on factual accuracy.",
    timeout=30,
)

# 2. PDF 변환기 설정 (이미지 캡션, 추출 켜기)
pipeline_options = PdfPipelineOptions()
pipeline_options.do_picture_description = True
pipeline_options.picture_description_options = openai_vlm_options
pipeline_options.images_scale = 2.0  # 이미지 크기 스케일업
pipeline_options.generate_picture_images = True  # 이미지 추출 파일 생성
pipeline_options.enable_remote_services = True

converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)
    }
)

# 3. 변환 및 이미지와 캡션 정보 저장
result = converter.convert("data/국가별 공공부문 AI 도입 및 활용 전략 (1).pdf")
doc = result.document

# --- 이미지 저장 방법 예시 ---
import base64
from PIL import Image
from io import BytesIO

output_img_dir = "parsed_images"
os.makedirs(output_img_dir, exist_ok=True)

# Docling 문서에서 picture 객체 추출
# 최신 패키지라면: doc.pictures, 아니면 doc.document.pictures 일수 있음
pictures = getattr(doc, "pictures", None)
if pictures is None:
    # fallback 경로
    pictures = getattr(getattr(doc, "document", None), "pictures", [])

for idx, picture in enumerate(pictures):
    try:
        # Url 객체를 문자열로 변환 후 base64 부분 추출
        uri_str = str(picture.image.uri)
        if ',' in uri_str:
            base64_data = uri_str.split(',')[1]

            # base64 디코딩 및 PIL 이미지로 변환
            image_data = base64.b64decode(base64_data)
            image = Image.open(BytesIO(image_data))

            # 파일명을 page_no 기준 + 순번으로
            page_no = getattr(picture.prov[0], "page_no", 0) if picture.prov else 0
            filename = f"page_{page_no}_img_{idx+1}.png"
            image.save(os.path.join(output_img_dir, filename))
            print(f"✅ Saved image: {filename}")
    except Exception as e:
        print(f"⚠️ Failed to save image {idx}: {e}")

# 4. 이미지 캡션/주석을 텍스트로 저장 (예: captions.txt)
captions_path = os.path.join(output_img_dir, "captions.txt")
with open(captions_path, "w", encoding="utf-8") as f:
    for idx, picture in enumerate(pictures):
        # caption_text는 doc 객체 필요
        caption = getattr(picture, "caption_text", None)
        if callable(caption):
            text = caption(doc=doc)
        else:
            text = getattr(picture, "caption", "")
        f.write(
            f"Picture #{idx}\n"
            f"Caption: {text}\n"
            f"Annotations: {getattr(picture, 'annotations', '')}\n\n"
        )
print(f"✅ Captions and annotations saved to {captions_path}")
```

```python
from docling_core.types.doc import PictureItem

output_dir = "picture_captions"
os.makedirs(output_dir, exist_ok=True)

captions_path = os.path.join(output_dir, "captions.txt")
with open(captions_path, "w", encoding="utf-8") as f:
    for element, _level in doc.iterate_items():
        if isinstance(element, PictureItem):
            f.write(
                f"Picture {element.self_ref}\n"
                f"Caption: {element.caption_text(doc=doc)}\n"
                f"Annotations: {element.annotations}\n\n"
            )
print(f"✅ Captions and annotations saved to {captions_path}")
```
