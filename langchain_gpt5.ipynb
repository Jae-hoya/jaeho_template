````python
llm = ChatOpenAI(model="gpt-5-mini")

llm.invoke("Hello, world!")

````

### reasoning
- "effort": "medium",  / `'low'`, `'medium'`, or `'high'`. / `'minimal'`
- "summary": "auto",  / `'detailed'`, `'auto'`, or `'None'`.

### verbosity
- Currently supported values are /`'low'`, `'medium'`, and `'high'`.

### temperature
- temperature, top_p, logprobs는 GPT-5에서(5.2) reasoning={"effort": None},일 때만 지원됩니다.

````python
model = ChatOpenAI(
    model="gpt-5-mini",
    output_version="responses/v1",
    reasoning={"effort": "medium"}, 
    verbosity="low",
)
model.invoke("hi")
````

````python
model = ChatOpenAI(
    model="gpt-5-mini",
    output_version="responses/v1",
    reasoning={"effort": "minimal"}, # low, medium, high 모두 가느느
    verbosity="low",
)
model.invoke("hi")
````

````python
model = ChatOpenAI(
    model="gpt-5-mini",
    output_version="responses/v1",
    reasoning={"effort": None},
    verbosity="high",
    temperature=0,
)
model.invoke("hi")
````
