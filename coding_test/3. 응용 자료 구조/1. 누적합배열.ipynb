# 누적합 배열

a[n]가 있으면 그 a[n+1] 만큼 더하는것.

배열 a의 연속한 구간 합

어디에 쓰느냐? 배열 a의 연속한 구간 합. 즉 구간합을 빠르게 구하기 위한 누적합

예)

1. 수열 a[2] + a[3]... a[8]: 시간 복잡도 O(N)

2. 누적합 배열 psum[8] - psum[1]: 시간복잡도 O(1)


# 2559: 수열
https://www.acmicpc.net/problem/2559

```python
# psum[i] = psum[i-1] + A[i]  누적합의 핵심
```

```python
N = list(map(int, input().split())) # 온도를 측정한 전체 날짜의 수수

K = list(map(int, input().split())) # 합을 구하기 위한 연속된 날짜의 수
# n일동안 측정한 온도 N개가 빈칸을 사이에 두고 주어진다. ????
A = list(map(int, input().split())) # 아! N 안의 수가? 주어진건가? 배열 N 안의 수(온도)!!


# 누적합: A[0]부터 A[N]까지의 합.
# A는 i번쨰로 날의 온도값
psum = [0] * N
psum[0] = A[0] 
for i in range(1, N): # 1~ n-1까지
    psum[i] = psum[i-1] + A[i] # 실질적으로 A[i]는 N이다

# 구간합: 연속된 K일은 K일의 시작 날짜를 기준으로 탐색한다. 시작날짜는 최대 n-k가 될 수 있다.
tem_sum = []
for i in range(0, N-K+1): # 마지막 배열의 인덱스는 N-1 0부터 n-k개까지 하므로 +1, 길이는 항상 K, i는 시작 구간 
    # i ~ i + K -1 K의 길이를 정의
    # # i가 시작일이므로 for문 내에서 i일째 부터 i+k-1일 째의 온도의 합을 for문안에서 구해야한다.
    if i == 0:
        sum = psum[i+K-1]
    else:
        sum = psum[i+K-1] - psum[i-1]
        
    tem_sum.append(sum)
    
print(max(tem_sum))




# 1 < K < N
# # 10일간 온도의 합이 가장 큰 값
# # N 은 날짜 수
# # K는 합을 구하기 위한 연속적인 날짜의 수

# max = 0

# for i in range(N):
    
#     for j in range(K):
#         sum += N(i+j)
        
#         if sum >= max:
#             max = sum
            
#         else:
#             pass
    
#             return max
```

# 14465: 수열
https://www.acmicpc.net/problem/14465

```python
N, K, B = list(map(int, input().split())) # 횡단보도
# N: 횡단보도
# K: 연속된 K개의 신호등
# B: 고장난 신호등

# 고장난 신호등에 번호가 주어진다. 고장나면 1
check = [0] * N #고장난 체크배열

# 체크 배열 안에 고장난 신호등의 위치를 입력을 받아 처리
for _ in range(B):
    id = int(input())
    check[id - 1] = 1

# 누적합    
psum = [0] * N # 누적합 체크배열
psum[0] = check[0]
for i in range(1,N):
    psum[i] = psum[i-1] + check[i] # 실질적으로 check[i]는 N이다

fix = [] # 고쳐야 하는 신호등 개수
for i in range(0, N - K +1): # 마지막 배열의 인덱스는 N-1 0부터 n-k개까지 하므로 +1, 길이는 항상 K, i는 시작 구간. 
    # i ~ i + K -1 : 구간의 합
    if i == 0:
        num = psum[i + K -1]
    else:
        num = psum[i + K -1] - psum[i-1]
        
    fix.append(num)
    
print(min(fix))
# 10개의 횡단보도가 있고 5개가 고장났다면, 6개의 연속된 신호등이 연결될라면 하나를 고쳐한다.
# 각 구간마다 몇개를 고쳐야 하는지
# 체크배열을 만들고, 각 위치에 고장났으면1, 멀쩡하면 0으로 처리
# 구간의 합이 가장 작은 값을 찾아야 한다. 이것은 누적합으로 구한다.
```
