```python
for i in range(1, 101):
    # {i:3}는 i를 3칸의 공간으로 오른쪽 정렬하여 출력한다는 의미입니다.
    # 즉, 1은 '  1', 10은 ' 10', 100은 '100'처럼 출력됩니다.
    print(f'i ={i:3} id(i)={id(i)}')
   # {i:<3}는 i를 3칸의 공간으로 왼쪽 정렬하여 출력한다는 의미입니다.
    # 즉, 1은 '  1', 10은 ' 10', 100은 '100'처럼 출력됩니다.
    print(f'i ={i:<3} id(i)={id(i)}')

    # {i:<3}는 i를 3칸의 공간으로 가운데 정렬하여 출력한다는 의미입니다.
    # 즉, 1은 '  1', 10은 ' 10', 100은 '100'처럼 출력됩니다.
    print(f'i ={i:^3} id(i)={id(i)}')
```

```python
x = [1, 2, 3, 4, 5]

a, *rest = x           # a=1, rest=[2,3,4,5]
print(a, rest)
*a, b = x              # a=[1,2,3,4], b=5
print(a, b)
a, *mid, b = x         # a=1, mid=[2,3,4], b=5
print(a, mid, b)
    
# 일부 무시하고 싶을 때 관례적으로 _ 사용
a, _, c = [10, 20, 30] # a=10, c=30
print(a, c)
a, *_, c = [10, 20, 30, 40, 50]  # a=10, c=50
print(a, c)
```

```python
a = list(map(int, input().split()))

def max_of(a):
    maximum = a[0]
    for i in range(1, len(a)):
        if a[i] > maximum:
            maximum = a[i]
    return maximum

max_of(a)
```

```python
print(max_of([1, 5, 3, 7, 2]))  # 7
print(max_of([-10, -2, -19]))   # -2
```

```python
# asd 함수는 전달된 리스트의 길이를 절반(n//2) 만큼 반복합니다.
# 각 반복에서 인덱스(i)와 전체 길이(n)를 출력하죠.
# 실제로 코드를 보면, n이 5일 때 i는 0, 1까지만 반복합니다.
def asd(a):
    n = len(a)
    for i in range(n//2):
        print(f"i: {i}, n: {n}")

asd([1,2,3,4,5])
```

```python
def reverse(a):
    n = len(a)
    for i in range(n//2):
        a[i], a[n-i-1] = a[n-i-1], a[i]


print(reverse([1,2,3,4,5]))
```

```python
def card_conv(x: int, r: int) -> str:
    d = ''
    dchar = '0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ'
    while x > 0:
        d += dchar[x % r]
        x //= r
    return d[::-1]
```

```python
def change(lst, idx, val):
    lst[idx] = val

x = [11,22,33,44,55]
print('x =', x)

index = int(input('업데이트할 인덱스 입력'))
value = int(input('새로운 값 입력'))

change(x, index, value)
print(f'x={x}')
```
