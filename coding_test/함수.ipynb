```python
# hassattr
# hasattr(obj, name)

# 기능: 객체 obj에 문자열 이름 name의 속성(메서드 포함)이 있으면 True, 없으면 False.
```

```python
class A:
    def foo(self): pass

a = A()
print("foo",hasattr(a, "foo"))    # True
print("bar",hasattr(a, "bar"))    # False
```
