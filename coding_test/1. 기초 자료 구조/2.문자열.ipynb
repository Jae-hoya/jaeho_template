### tip
1. 문장이 주어졌을때 문장에 나오는 알파벳 개수 각각 알고싶은 상황이 있다고 하면, "codingtest"에 a는 몇개? b는 몇개... z는 몇개인지 알고싶을 때?

`count`배열을 만들어서 사용: a가 나왔다 하면, count배열에 0번째 index +1, b가나오면 1번째 index +1, z가 나오면 25 index +1 -> 함수를 만들어야한다. dict와 같은 객체를 생성

2. 'b' - 'a' 가 설립할 때, 이것은 1이 나온다 'c' - 'a' = 2가 나온다.... 'z' - '1' = 25가 나온다. 
character들은 그에 대응하는 숫자들이 정해져 있기 때문이다(아스키 코드)


# 10809 알파벳 찾기
https://www.acmicpc.net/problem/10809

```python
# 문제
# 알파벳 소문자로만 이루어진 단어 S가 주어진다.
# 각각의 알파벳에 대해서, 단어에 포함되어 있는 경우에는 처음 등장하는 위치를, 
# 포함되어 있지 않은 경우에는 -1을 출력하는 프로그램을 작성하시오.

# 입력
# 첫째 줄에 단어 S가 주어진다. 단어의 길이는 100을 넘지 않으며, 알파벳 소문자로만 이루어져 있다.

# 출력
# 각각의 알파벳에 대해서, a가 처음 등장하는 위치, b가 처음 등장하는 위치, ... z가 처음 등장하는 위치를 공백으로 구분해서 출력한다.

# 만약, 어떤 알파벳이 단어에 포함되어 있지 않다면 -1을 출력한다. 단어의 첫 번째 글자는 0번째 위치이고, 두 번째 글자는 1번째 위치이다.
```

```python
s = input()

check = [-1] * 26

for i in range(len(s)): # 문자열
    index = ord(s[i]) - ord('a') # 아스키 코드를 구하고 소문자 a의 아스키 코드(0)를 빼서 0~25 사이의 숫자로 만든다.
    
    # i 는 -1경우에만 바꿔준다
    if check[index] == -1:
        check[index] = i
        
    else:
        pass
    
for i in range(26):
    print(check[i], end = ' ')
```

# 11720 숫자의 합
https://www.acmicpc.net/problem/11720

```python
# 문제
# N개의 숫자가 공백 없이 쓰여있다. 이 숫자를 모두 합해서 출력하는 프로그램을 작성하시오.

# 입력
# 첫째 줄에 숫자의 개수 N (1 ≤ N ≤ 100)이 주어진다. 둘째 줄에 숫자 N개가 공백없이 주어진다.

# 출력
# 입력으로 주어진 숫자 N개의 합을 출력한다.
```

```python
# 문자 9를 숫자 9로 바꿔야 하는데, 이것은 아스키 코드를 사용한다
# ord('9') - ord('0') = 9
# ord('8') - ord('0') = 8


n = int(input())
s = input() # 문자열

sum = 0
for i in range(n):
    sum += ord(s[i]) - ord('0') # '9' - '0' = 9

print(sum)

```
