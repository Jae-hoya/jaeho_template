# 스텍

### 자료구조
데이터를 어떻게 저장하고 관리 할까에 대한 방법론.

단순히 저장 공간이 아닌, 어떻게 저장하고 관리할지에 대한 방법론이다.

## 스택 & 큐
데이터가 추가/ 제거될떄 어떤 규칙을 가지고 하는지에 대한 관리 방법이다

### 스택
LIFO: Last In First Out: 마지막에 들어온것을 우선해서 제거한다

예를들어: 5, 4, 3, 제거, 제거 가 수행된다면, 5만 남는다.

```python
# pop() 함수는 스택(또는 리스트)에서 마지막 요소를 꺼내서 반환하는 함수이다.
# 즉, 마지막에 추가된 데이터를 제거할 때 사용한다.
# 예시:
# stack = [1, 2, 3]
# stack.pop()  # 3을 반환하고, stack은 [1, 2]가 된다.
#
# pop() 함수에 인자로 -1, 0, 1 등을 넣으면 해당 인덱스의 요소를 꺼낸다.
# stack.pop(-1)은 마지막 요소(3)를, stack.pop(0)은 첫 번째 요소(1), stack.pop(1)은 두 번째 요소(2)를 꺼낸다.
# 하지만, 스택 자료구조에서는 마지막에 들어온 데이터를 꺼내는 것이 원칙(LIFO)이므로,
# 항상 pop() 또는 pop(-1)을 사용해서 마지막 요소를 꺼내는 것이 바람직하다.
# 예를 들어, stack.pop(0)이나 stack.pop(1)을 사용하면 스택의 원칙에 어긋난다.
```

# 10773: 제로
https://www.acmicpc.net/problem/10773

```python
# 문제
# 나코더 기장 재민이는 동아리 회식을 준비하기 위해서 장부를 관리하는 중이다.

# 재현이는 재민이를 도와서 돈을 관리하는 중인데, 애석하게도 항상 정신없는 재현이는 돈을 실수로 잘못 부르는 사고를 치기 일쑤였다.

# 재현이는 잘못된 수를 부를 때마다 0을 외쳐서, 가장 최근에 재민이가 쓴 수를 지우게 시킨다.

# 재민이는 이렇게 모든 수를 받아 적은 후 그 수의 합을 알고 싶어 한다. 재민이를 도와주자!

# 입력
# 첫 번째 줄에 정수 K가 주어진다. (1 ≤ K ≤ 100,000)

# 이후 K개의 줄에 정수가 1개씩 주어진다. 정수는 0에서 1,000,000 사이의 값을 가지며, 
# 정수가 "0" 일 경우에는 가장 최근에 쓴 수를 지우고, 아닐 경우 해당 수를 쓴다.

# 정수가 "0"일 경우에 지울 수 있는 수가 있음을 보장할 수 있다.

# 출력
# 재민이가 최종적으로 적어 낸 수의 합을 출력한다. 최종적으로 적어낸 수의 합은 231-1보다 작거나 같은 정수이다.
```

```python
# 잘못된 숫자를 부를때 0을 불러, 재민이 쓴 수를 지운다.
# not 0 : 사용한 돈의 의미: 기록을 추가
# 0: 실수 : 기록을 지워라
# 장부에 기록되어있는 순자를 합산하여 출력


k = int(input())

stack = []

for i in range(k):
    n = int(input())
    
    if n == 0:
        stack.pop(-1) # 미자믹 추가된 데이터를 삭제
    else:
        stack.append(n) # 숫자 n을 추가

sum = 0
for e in stack:
    sum += e
    
print(sum)
    
```

# 4949: 균형잡힌 세상
https://www.acmicpc.net/problem/4949

```python
# 문제
# 세계는 균형이 잘 잡혀있어야 한다. 양과 음, 빛과 어둠 그리고 왼쪽 괄호와 오른쪽 괄호처럼 말이다.

# 정민이의 임무는 어떤 문자열이 주어졌을 때, 괄호들의 균형이 잘 맞춰져 있는지 판단하는 프로그램을 짜는 것이다.

# 문자열에 포함되는 괄호는 소괄호("()") 와 대괄호("[]")로 2종류이고, 문자열이 균형을 이루는 조건은 아래와 같다.

# 모든 왼쪽 소괄호("(")는 오른쪽 소괄호(")")와만 짝을 이뤄야 한다.
# 모든 왼쪽 대괄호("[")는 오른쪽 대괄호("]")와만 짝을 이뤄야 한다.
# 모든 오른쪽 괄호들은 자신과 짝을 이룰 수 있는 왼쪽 괄호가 존재한다.
# 모든 괄호들의 짝은 1:1 매칭만 가능하다. 즉, 괄호 하나가 둘 이상의 괄호와 짝지어지지 않는다.
# 짝을 이루는 두 괄호가 있을 때, 그 사이에 있는 문자열도 균형이 잡혀야 한다.
# 정민이를 도와 문자열이 주어졌을 때 균형잡힌 문자열인지 아닌지를 판단해보자.

# 입력
# 각 문자열은 마지막 글자를 제외하고 영문 알파벳, 공백, 소괄호("( )"), 대괄호("[ ]")로 이루어져 있으며, 
# 온점(".")으로 끝나고, 길이는 100글자보다 작거나 같다.

# 입력의 종료조건으로 맨 마지막에 온점 하나(".")가 들어온다.
# 출력
# 각 줄마다 해당 문자열이 균형을 이루고 있으면 "yes"를, 아니면 "no"를 출력한다.
```

```python
# 괄호를 올바르게 잘 찾느냐에 대한 문제

# 온점이 올떄까지 True
while True:
    s = input()
    
    stack = []
    balanced = True
    
    for i in range(len(s)):
        if s[i] == '(' or s[i] == '[':
            stack.append(s[i])
        
        if s[i] == ')':
            # stack이 비어있으면 안됌
            if len(stack) == 0:
                balanced = False
                break
            
            last = stack.pop(-1) # stack의 마지막 문자열
            if last != '(':
                balanced = False
                break
                
        if s[i] == ']':
            # stack이 비어있으면 안됌
            if len(stack) == 0:
                balanced = False
                break
            
            last = stack.pop(-1) # stack의 마지막 문자열
            if last != '[':
                balanced = False
                break
    
    if s == ".":
        break
    
# 마지막에는 스텍이 비어있어야 한다.
# 괄호다 닫힌다는 뜻이, stack이 비어있다는 뜻이다.
if len(stack) != 0:
    balanced = False
    
if balanced and len(stack) == 0:
    print("yes")
else:
    print("no")

```
