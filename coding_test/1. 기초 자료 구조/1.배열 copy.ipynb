# 5597 과제 안 내신 분..?

https://www.acmicpc.net/problem/5597

```python
# 문제
# X대학 M교수님은 프로그래밍 수업을 맡고 있다. 교실엔 학생이 30명이 있는데, 학생 명부엔 각 학생별로 1번부터 30번까지 출석번호가 붙어 있다.

# 교수님이 내준 특별과제를 28명이 제출했는데, 그 중에서 제출 안 한 학생 2명의 출석번호를 구하는 프로그램을 작성하시오.

# 입력
# 입력은 총 28줄로 각 제출자(학생)의 출석번호 n(1 ≤ n ≤ 30)가 한 줄에 하나씩 주어진다. 출석번호에 중복은 없다.

# 출력
# 출력은 2줄이다. 1번째 줄엔 제출하지 않은 학생의 출석번호 중 가장 작은 것을 출력하고, 2번째 줄에선 그 다음 출석번호를 출력한다.
```

```python
check = [0] * 31 # 31개의 체크배열, 0은 사용 x

for i in range(28): # 28명 제출. 0~27
    n = int(input()) # input 입력받음
    check[n] = 1 # 입력받은 체크배열은 1
    
answer = [] # 정답을 저장할 배열. 
for i in range(1, 31): # 1~30
    if check[i] == 0: # 체크배열이 0이면 정답에 추가
        answer.append(i)
        
print(answer[0])
print(answer[1])
```

# 3062 나머지 
https://www.acmicpc.net/problem/3052

```python
# 문제
# 두 자연수 A와 B가 있을 때, A%B는 A를 B로 나눈 나머지 이다. 예를 들어, 7, 14, 27, 38을 3으로 나눈 나머지는 1, 2, 0, 2이다. 

# 수 10개를 입력받은 뒤, 이를 42로 나눈 나머지를 구한다. 그 다음 서로 다른 값이 몇 개 있는지 출력하는 프로그램을 작성하시오.

# 입력
# 첫째 줄부터 열번째 줄 까지 숫자가 한 줄에 하나씩 주어진다. 이 숫자는 1,000보다 작거나 같고, 음이 아닌 정수이다.

# 출력
# 첫째 줄에, 42로 나누었을 때, 서로 다른 나머지가 몇 개 있는지 출력한다.
```

```python
answer = []
for i in range(100):
    if i % 42 != 0:
        answer.append(i)

print(answer, "\n", len(answer))
```

```python
# 체크배열.
check = [0] * 42

for i in range(10): # 입력 0 - 9
    n = int(input())
    
    check[n % 42] = 1 # 입력받은 n을 42로 나누고 나머지를 구한다.
    
sum = 0 # 종류의 개수 = 배열의 총합
for i in range(42): # 0 - 41
    sum += check[i]

print(sum)

# 42로 나눈 나머지 체크배열은 인덱스가 0(최소)부터 41(최대)까지 즉, 최대값은 41.
# 중복된 나머지는 1로 덮어쓰기 때문에, 체크배열 총합이 서로 다른 나머지의 개수와 같음
```

```python
for i in range(28): # 0 - 28
    print(i)
```
