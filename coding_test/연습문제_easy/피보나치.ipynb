```python
def solution(n: int) -> int:
    # 제약: 0 <= n <= 30  (필요하면 주석으로 표시)
    if n == 0:
        return 0
    a = [0, 1]  # 초기값
    for i in range(2, n + 1):           # range를 사용
        a.append(a[i-1] + a[i-2])       # 올바른 점화식
    return a[n]                          # answer 대신 a[n]


solution(7)
```

```python
# 두 피보나치 값의 곱 구하기
def solution(n:int, m:int):
    if not (0<=n<=30 and 0<=m<=30):
        raise ValueError

    a = [0, 1]
    b = [0, 1]

    for i in range (2, n+1):
        a.append(a[i-1] + a[i-2])
    
    for o in range (2, m+1):
        b.append(b[o-1] + b[o-2])
       
    return a[n] * b[m]
```

```python
# 두 피보나치 수열의 홀 짝 구하기
def solution_verbose(n: int, m: int) -> bool:
    if not (0 <= n <= 30 and 0 <= m <= 30):
        raise ValueError("n, m은 0~30 사이 정수여야 합니다.")

    a = [0, 1]
    b = [0, 1]

    for i in range(2, n + 1):
        a.append(a[i - 1] + a[i - 2])

    for o in range(2, m + 1):
        b.append(b[o - 1] + b[o - 2])

    prod = a[n] * b[m]
    return (prod % 2) == 1   # 홀수면 True
```
