```python
# 4명이 탁구각각 스킬레벨을 가지고 있다.
# 두명씩 나누어 팀을 한다.
# 각각 최대한 비슷한 스킬레벨을 가지는 팀을 만들어야 한다.
# 두 팀의 스킬레벨 차이가 최소가 되도록 하려고 한다.

A, B, C, D = map(int, input().split())

diff = 10000000

# abs 평균

diff = min(diff, abs(A+B-C-D))
diff = min(diff, abs(A+C-B-D))
diff = min(diff, abs(A+D-B-C))

# 4 7 10 20

print(diff)
```
