```python
# (월,일) -> 연중 몇 번째 날(1~365)로 변환 (윤년 아님)
def month_day_to_day_number(month: int, day: int) -> int:
    days_in_month = [31, 28, 31, 30, 31, 30,
                     31, 31, 30, 31, 30, 31]
    # 간단 유효성(월 범위, 일 범위) 체크
    return sum(days_in_month[:month - 1]) + day # 현재의 달 전까지의 일수 + 현재의 일수 몇번째 일수인지 알려준다. 
    # 정확히는 직전달까지 보기 위해서 -1이다. 그러니까, 달을 연중일수로 생각해서 해결한것이다.

# 기존 weekend 계산을 확장: holidays를 선택 인자로 받음
def solution(x: int, holidays: [list[[int, int]]]):
    assert 1 <= x <= 7, "x는 1(월)~7(일)이어야 합니다."

    # 1) 주말(토=6, 일=7)인 연중 날짜 집합 만들기
    weekend_days: set[int] = set()
    for day in range(1, 366):  # 1..365
        weekday = ((x - 1) + (day - 1)) % 7 + 1  # day=1일 때 weekday=x
        if weekday in (6, 7):
            weekend_days.add(day)

    # 2) 공휴일이 주어지면 day-of-year로 변환해서 합집합
    if holidays:
        holiday_days: set[int] = set()
        for m, d in holidays:
            dn = month_day_to_day_number(m, d)
            if 1 <= dn <= 365:  # 잘못된 날짜(-1 등)는 무시
                holiday_days.add(dn)
        weekend_days = weekend_days.union(holiday_days)# 합집합: 겹치면 한 번만 카운트

    return len(weekend_days)

print(solution(7, [(3,3)]))
```

```python
# 공휴일 정보가 주어질 때, 휴일이 총 며칠 있는지 구한다.
# 공휴일 달력에서, 노라색으로 칠해진 부분. 원래 쉬었던 주말과 겹치치 않는 공휴일은 10일 이믈 115일을 쉰다.

from typing import List, Tuple

def month_day_to_day_number(month: int, day: int) -> int:
    days_in_month = [31, 28, 31, 30, 31, 30,
                     31, 31, 30, 31, 30, 31]
    return sum(days_in_month[:month - 1]) + day


def get_weekend_days(start_day: int) -> set:
    weekend_days = set()
    for day in range(1, 366):
        weekday = (start_day + day - 2) % 7 + 1
        if weekday in [6, 7]:
            weekend_days.add(day)
    return weekend_days


def get_holiday_days(holidays: List[Tuple[int, int]]) -> set:
    holiday_days = set()
    for month, day in holidays:
        day_number = month_day_to_day_number(month, day)
        if 1 <= day_number <= 365:
            holiday_days.add(day_number)
    return holiday_days


def solution(x: int, holidays: List[Tuple[int, int]]) -> int:
    weekends = get_weekend_days(x)
    holiday_days = get_holiday_days(holidays)
    total_rest_days = weekends.union(holiday_days)
    return len(total_rest_days)



# 임시 공휴일을 주어진 규칙대로 지정한다.
from typing import List, Tuple, Set

def month_day_to_day_number(month: int, day: int) -> int:
    days_in_month = [31, 28, 31, 30, 31, 30,
                     31, 31, 30, 31, 30, 31]
    return sum(days_in_month[:month - 1]) + day


def get_weekend_days(start_day: int) -> Set[int]:
    weekends = set()
    for day in range(1, 366):
        weekday = (start_day + day - 2) % 7 + 1
        if weekday in [6, 7]:  # 토, 일
            weekends.add(day)
    return weekends


def get_holiday_days(holidays: List[Tuple[int, int]]) -> Set[int]:
    result = set()
    for m, d in holidays:
        day_number = month_day_to_day_number(m, d)
        if 1 <= day_number <= 365:
            result.add(day_number)
    return result


def apply_substitute_holidays(holiday_days: Set[int], weekend_days: Set[int], start_day: int) -> Set[int]:
    substitute_days = set()

    for day in holiday_days:
        weekday = (start_day + day - 2) % 7 + 1

        # 토요일 → 금요일 대체
        if weekday == 6:
            if (day - 1) >= 1 and (day - 1) not in weekend_days and (day - 1) not in holiday_days:
                substitute_days.add(day - 1)

        # 일요일 → 월요일 대체
        elif weekday == 7:
            if (day + 1) <= 365 and (day + 1) not in weekend_days and (day + 1) not in holiday_days:
                substitute_days.add(day + 1)

    return substitute_days


def solution(x: int, holidays: List[Tuple[int, int]]) -> int:

    weekend_days = get_weekend_days(x)
    holiday_days = get_holiday_days(holidays)
    substitute_days = apply_substitute_holidays(holiday_days, weekend_days, x)

    total_rest_days = weekend_days.union(holiday_days).union(substitute_days)

    return len(total_rest_days)

# 주어진 규칙에 따라 월급날이 언제로 조정되는지 구한다.

# 월급날이 평일인 경우 병호는 그날 월급을 받는다
# 월급날이 공휴일 또는 주말인 겨웅 병호는 가장 가까운 평일에 월급을 받는다
# 2번 조건에 해당하는 일이 여러개인 경우, 더 이른 날짜에 월급을 받는다.
# 월월급은 전 달이나 다음달로 넘어가서 받지 않으며, 해당 월에조정 가능한 날이 없다면 y일에 받는다.

# Solution.py
from typing import List, Tuple, Set

# --- 유틸: 날짜 변환 ---
def month_day_to_day_number(month: int, day: int) -> int:
    days_in_month = [31, 28, 31, 30, 31, 30,
                     31, 31, 30, 31, 30, 31]
    return sum(days_in_month[:month - 1]) + day

def month_bounds(month: int) -> Tuple[int, int, int]:
    days_in_month = [31, 28, 31, 30, 31, 30,
                     31, 31, 30, 31, 30, 31]
    start_dn = sum(days_in_month[:month - 1]) + 1
    end_dn = start_dn + days_in_month[month - 1] - 1
    return start_dn, end_dn, days_in_month[month - 1]

# --- 주말/공휴일/대체공휴일 계산 ---
def get_weekend_days(start_day: int) -> Set[int]:
    """start_day: 1=월 … 7=일 기준의 1/1 요일"""
    weekends = set()
    for day in range(1, 366):
        weekday = (start_day + day - 2) % 7 + 1
        if weekday in (6, 7):  # 토, 일
            weekends.add(day)
    return weekends

def get_holiday_days(holidays: List[Tuple[int, int]]) -> Set[int]:
    result = set()
    for m, d in holidays:
        dn = month_day_to_day_number(m, d)
        if 1 <= dn <= 365:
            result.add(dn)
    return result

def apply_substitute_holidays(holiday_days: Set[int],
                              weekend_days: Set[int],
                              start_day: int) -> Set[int]:
    """토요일 공휴일→전날(금), 일요일 공휴일→다음날(월)로 대체. 
       단, 그 날이 또 휴일/주말이면 대체하지 않음."""
    substitute_days = set()
    for dn in holiday_days:
        weekday = (start_day + dn - 2) % 7 + 1
        # 토요일 → 금요일
        if weekday == 6:
            prev_dn = dn - 1
            if prev_dn >= 1 and prev_dn not in weekend_days and prev_dn not in holiday_days:
                substitute_days.add(prev_dn)
        # 일요일 → 월요일
        elif weekday == 7:
            next_dn = dn + 1
            if next_dn <= 365 and next_dn not in weekend_days and next_dn not in holiday_days:
                substitute_days.add(next_dn)
    return substitute_days

def build_rest_days(start_day: int, holidays: List[Tuple[int, int]]) -> Set[int]:
    weekend_days = get_weekend_days(start_day)
    holiday_days = get_holiday_days(holidays)
    substitute_days = apply_substitute_holidays(holiday_days, weekend_days, start_day)
    return weekend_days | holiday_days | substitute_days

# --- 월급일 조정 규칙 ---
def adjust_payday_for_month(month: int,
                            payday: int,
                            fallback_y: int,
                            start_day: int,
                            holidays: List[Tuple[int, int]]) -> int:
    """
    규칙:
    1) 월급날이 평일이면 그대로.
    2) 공휴일/주말이면 같은 달 안에서 '가장 가까운 평일'로 이동(거리가 같으면 '더 이른 날짜' 우선).
    3) 월을 넘기지 않음. 같은 달에서 조정 불가 시 해당 월의 y일에 지급.
    """
    rest = build_rest_days(start_day, holidays)
    start_dn, end_dn, mdays = month_bounds(month)

    # 월 범위 보정
    payday = min(max(1, payday), mdays)
    fallback_y = min(max(1, fallback_y), mdays)

    dn = month_day_to_day_number(month, payday)

    # 1) 평일이면 그대로
    if dn not in rest:
        return payday

    # 2) 같은 달 안에서 가장 가까운 평일 탐색(동일 거리면 더 이른 날짜 우선)
    dist = 1
    while True:
        prev_dn = dn - dist
        next_dn = dn + dist

        prev_ok = (prev_dn >= start_dn) and (prev_dn not in rest)
        next_ok = (next_dn <= end_dn) and (next_dn not in rest)

        if prev_ok and next_ok:
            chosen_dn = prev_dn  # 더 이른 날짜 우선
            break
        elif prev_ok:
            chosen_dn = prev_dn
            break
        elif next_ok:
            chosen_dn = next_dn
            break

        if prev_dn < start_dn and next_dn > end_dn:
            chosen_dn = None
            break

        dist += 1

    # 3) 조정 불가 → y일
    if chosen_dn is None:
        return fallback_y

    # day-number → day
    adjusted_day = chosen_dn - (start_dn - 1)
    return adjusted_day

def adjust_paydays_for_year(start_day: int,
                            holidays: List[Tuple[int, int]],
                            payday: int,
                            fallback_y: int) -> List[int]:
    return [
        adjust_payday_for_month(m, payday, fallback_y, start_day, holidays)
        for m in range(1, 12 + 1)
    ]

# Solution.py (핵심 엔트리포인트만 교체)
from typing import List, Tuple

# ... (여기에 기존의 month_day_to_day_number, month_bounds,
#      get_weekend_days, get_holiday_days, apply_substitute_holidays,
#      build_rest_days, adjust_payday_for_month, adjust_paydays_for_year 가 그대로 있어야 합니다)

def solution(x: int, holidays: List[Tuple[int, int]], *args):
    """
    다형 엔트리포인트:
    - solution(x, holidays) -> int
        연간 쉬는 날(주말 ∪ 공휴일 ∪ 대체공휴일) 총일수 반환
    - solution(x, holidays, payday) -> List[int]
        각 월의 조정된 월급일 반환 (fallback_y는 payday로 가정)
    - solution(x, holidays, payday, fallback_y) -> List[int]
        각 월의 조정된 월급일 반환
    """
    # 기본 검증
    if not (1 <= x <= 7):
        raise ValueError("x는 1(월)~7(일) 사이의 정수여야 합니다.")

    # 1) 인자 2개: 쉬는 날 총일수
    if len(args) == 0:
        rest = build_rest_days(x, holidays)
        return len(rest)

    # 2) 인자 3개: (x, holidays, payday)
    elif len(args) == 1:
        payday = int(args[0])
        if not (1 <= payday <= 31):
            raise ValueError("payday는 1~31 사이 정수여야 합니다.")
        # 같은 달에서 조정 불가 시 y일은 명시가 없으므로 payday로 가정
        fallback_y = payday
        return adjust_paydays_for_year(x, holidays, payday, fallback_y)

    # 3) 인자 4개: (x, holidays, payday, fallback_y)
    elif len(args) == 2:
        payday = int(args[0])
        fallback_y = int(args[1])
        return adjust_paydays_for_year(x, holidays, payday, fallback_y)

    else:
        raise TypeError("solution은 2, 3, 4개의 인자 형태만 지원합니다.")


# 병호가 사용할 수 있는 휴가 개수가 주어질 때, 가장 오래 쉴 수 있는 기간을 구한다.
# 병호가 휴가를 받은일 N은 15보다 작은 자연수이다.
# 병호가 연속적으로 쉴 수 있는 기간과 휴가 시작 월, 일 휴가 끝 월, 일을 차례로 반환한다. 답이 어러개라면 휴가 시작이 가장 빠른 답을 반환한다.
# 병호는 휴가 3일을 받았으므로, 9월 25, 9월 26, 10월 2일에 휴가를 사용하면 9월 23일부터 총 10월 3일까지 쉴 수 있다.

# Implement the single `solution` function (as previously provided) and run it
from typing import List, Tuple

def solution(start_day: int, holidays: List[Tuple[int, int]], N: int) -> Tuple[int, int, int, int, int]:
    """
    반환: (연속일수, 시작월, 시작일, 종료월, 종료일)
    - 쉬는 날: 주말(토/일), 공휴일, 대체공휴일(토 공휴일→전일, 일 공휴일→익일; 연쇄 대체 없음)
    - 병호는 휴가 N일을 근무일에 배치할 수 있음(1 ≤ N < 15)
    - 최장 길이가 동률이면 시작이 더 이른 구간 선택
    """
    if not (1 <= start_day <= 7):
        raise ValueError("start_day는 1(월)~7(일)이어야 합니다.")
    if not (1 <= N < 15):
        raise ValueError("N은 1 이상 14 이하의 자연수여야 합니다.")

    DAYS_IN_MONTH = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

    def month_day_to_day_number(month: int, day: int) -> int:
        return sum(DAYS_IN_MONTH[:month - 1]) + day

    def day_number_to_month_day(dn: int) -> Tuple[int, int]:
        m = 1
        while m <= 12:
            if dn > DAYS_IN_MONTH[m - 1]:
                dn -= DAYS_IN_MONTH[m - 1]
                m += 1
            else:
                return m, dn
        return 12, DAYS_IN_MONTH[-1]

    def get_weekend_days() -> set:
        weekends = set()
        for dn in range(1, 366):
            weekday = (start_day + dn - 2) % 7 + 1  # 1=Mon ... 7=Sun
            if weekday in (6, 7):  # Sat=6, Sun=7
                weekends.add(dn)
        return weekends

    def get_holiday_days() -> set:
        hs = set()
        for m, d in holidays:
            dn = month_day_to_day_number(m, d)
            if 1 <= dn <= 365:
                hs.add(dn)
        return hs

    def apply_substitute_holidays(holiday_days: set, weekend_days: set) -> set:
        subs = set()
        for dn in holiday_days:
            weekday = (start_day + dn - 2) % 7 + 1
            if weekday == 6:  # Sat -> Friday
                cand = dn - 1
                if cand >= 1 and cand not in weekend_days and cand not in holiday_days:
                    subs.add(cand)
            elif weekday == 7:  # Sun -> Monday
                cand = dn + 1
                if cand <= 365 and cand not in weekend_days and cand not in holiday_days:
                    subs.add(cand)
        return subs

    weekend_days = get_weekend_days()
    holiday_days = get_holiday_days()
    substitute_days = apply_substitute_holidays(holiday_days, weekend_days)
    rest_days = weekend_days | holiday_days | substitute_days

    is_rest = [False] * 366
    for dn in rest_days:
        is_rest[dn] = True

    best_len = 0
    best_l = 1
    l = 1
    work_cnt = 0

    for r in range(1, 366):
        if not is_rest[r]:
            work_cnt += 1
        while work_cnt > N and l <= r:
            if not is_rest[l]:
                work_cnt -= 1
            l += 1
        cur_len = r - l + 1
        if (cur_len > best_len) or (cur_len == best_len and l < best_l):
            best_len = cur_len
            best_l = l

    best_r = best_l + best_len - 1
    sm, sd = day_number_to_month_day(best_l)
    em, ed = day_number_to_month_day(best_r)
    return best_len, sm, sd, em, ed

# Given inputs (correcting the apparent typo "[,29]" to (9,29))
x = 7
n = 3
h = [(1,1), (1,21), (1,22), (1,23), (3,1), (5,5), (5,27), (6,6), (8,15), (9,28), (9,29), (9,30), (10,3), (10,9), (12,25)]

result = solution(x, h, n)
result
```
