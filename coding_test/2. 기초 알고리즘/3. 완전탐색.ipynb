# 완전 탐색
모든 경우의 수에 대하여 모두 시도해 보는 방법.

조합, 순열, 부분집합

- 조합
주머니에서 공 4개중 2개를 한번에 꺼내는 시나리오

- 순열
주머니에서 공 2개를 순서대로 꺼내는 시나리오

- 부분집합
주머니에서 공을 아무렇게나 한번에 꺼내는 시나리오 

4개의 원소의 부분집합

하나도 안뽑기도가능, 1개 2개, 3개, 4개를 뽑는 시나리오 가능

# 조합

```python
from itertools import combinations

pocket = [1, 2, 3,4]
list(combinations(pocket,2))
```

# 순열

```python
from itertools import permutations

pocket = [1, 2, 3, 4]
list(permutations(pocket,2))
```

# 부분집합

`append()`는 전체 객체를 하나의 요소로 취급하고, `extend()`는 객체 안의 개별 요소를 하나씩 추가합니다.

```python
from itertools import combinations

pocket = [1,2,3,4]
subset = []

for i in range(len(pocket)+1):
    subset.append(list(combinations(pocket, i)))
    
subset
```

```python
from itertools import combinations

pocket = [1,2,3,4]
subset = []

for i in range(len(pocket)+1):
    subset.extend(list(combinations(pocket, i)))
    
subset
```

# 2309 일곱난쟁이
https://www.acmicpc.net/problem/2309

```python
# 9명중 7명의 키의 합은 100
# 7명을 뽑아서 합이 100인지 확인해야한다.

from itertools import combinations # 조합

heights = []
for _ in range(9):
    height = int(input())
    heights.append(height)


for a in (combinations(heights, 7)):
    if sum(a) == 100:
        a = list(a)
        a.sort()
        for x in a:
            print(x)
        break
```

# 10819 차이를 최대로
https://www.acmicpc.net/problem/10819

```python
from itertools import permutations # 순열: 모든 가능한 순서를 모두 실험하기 때문

N = int(input()) # 정수
A = list(map(int, input().split())) # 배열

for a in permutations(A, N): # A에서 N개를 뽑는다.
    diff_sum = 0
    for i in range(N-1):
        diff_sum += abs(a[i] - a[i+1])
        
    max_diff_sum = max(diff_sum, diff_sum)
    
print(max_diff_sum)
```
