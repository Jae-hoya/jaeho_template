# 이분탐색

### 시간복잡도
+, -, *, /, =, >, < 와 같은 기본 연산들을 몇번을 했는지를 말한다. 여러가지 프로개름들 사이에서 우수성을 비교하기 위함이다.

이때, n이 매우 커졌을때 상관없는것들이 있으면 모두 생략하고 O(n)으로 상ㅇ한다

ex) 3n^2 + 9n + 3 -> 앞에 붙은 숫자들 다 날리고, n날리고 상수항 날리고 O(n^2)

ex) 9999n + 4  -> 앞에붙은 숫자, 상수항 다날리고 O(n)

### 이분탐색
- 답을 탐색하는 문제 -> 탐색의 범위를 반씩 줄여 나가는 방법

ex) 순서대로 정렬이 되어있는 길이 N 배열이 주어졌을 때, 특정 숫자 x가 있는지 찾는 문제.
답의 탐색 범위를 규명 [0, N-1]

쉬운방법 vs 이분탐색

- 쉬운방법
0부터 n-1까지 for문을 돌면서 n번 하게됌.. [1, N-1], ... , [N, N-1] 따라서 시간복잡도가 엄청 커진다.

- 이분탐색
골라야 하는 값이 y라 할때,
1. 가운데 인덱스를 골라서 확인한다 (x)
2. x > y, y = x, y < x 를 고려해야한다. 이때 y=x면 좋은거고, 나머지는 탐색의 범위를 날리는 것이다. 즉     log<sub>2</sub>n번을 한다. (N을 2로 몇번 나눌 수 있는지.) : O(log<sub>2</sub>n)

- O(N) vs O(log<sub>2</sub>n)
N=100이면 , O(N)은 100번, O(log<sub>2</sub>n)은 약 6번이면 된다.
N=10억이면 , O(N)은 10억번, O(log<sub>2</sub>n)은 약 30번이면 된다.

# 2512: 예산
https://www.acmicpc.net/problem/2512

```python
# 문제
# 국가의 역할 중 하나는 여러 지방의 예산요청을 심사하여 국가의 예산을 분배하는 것이다. 
# 국가예산의 총액은 미리 정해져 있어서 모든 예산요청을 배정해 주기는 어려울 수도 있다. 그래서 정해진 총액 이하에서 가능한 한 최대의 총 예산을 다음과 같은 방법으로 배정한다.

# 모든 요청이 배정될 수 있는 경우에는 요청한 금액을 그대로 배정한다.
# 모든 요청이 배정될 수 없는 경우에는 특정한 정수 상한액을 계산하여 그 이상인 예산요청에는 모두 상한액을 배정한다. 
# 상한액 이하의 예산요청에 대해서는 요청한 금액을 그대로 배정한다. 
# 예를 들어, 전체 국가예산이 485이고 4개 지방의 예산요청이 각각 120, 110, 140, 150이라고 하자. 
# 이 경우, 상한액을 127로 잡으면, 위의 요청들에 대해서 각각 120, 110, 127, 127을 배정하고 그 합이 484로 가능한 최대가 된다. 

# 여러 지방의 예산요청과 국가예산의 총액이 주어졌을 때, 위의 조건을 모두 만족하도록 예산을 배정하는 프로그램을 작성하시오.

# 입력
# 첫째 줄에는 지방의 수를 의미하는 정수 N이 주어진다. N은 3 이상 10,000 이하이다. 
# 다음 줄에는 각 지방의 예산요청을 표현하는 N개의 정수가 빈칸을 사이에 두고 주어진다. 
# 이 값들은 모두 1 이상 100,000 이하이다. 그 다음 줄에는 총 예산을 나타내는 정수 M이 주어진다. M은 N 이상 1,000,000,000 이하이다. 

# 출력
# 첫째 줄에는 배정된 예산들 중 최댓값인 정수를 출력한다. 
```

```python
# 485를  예산 120 110 140 150으로 나눠야 한다. 상한액을 127로 잡아, 120, 110, 127, 127로 배정하고 484가 된다.
# 상한액을 가장 크게 잡을 수 있는 프로그램을 작성해라

# 가운대 값을 구해서, 그 값을 기준으로 지방세를 비교한다.
# 이때, 이 지방세 합이 총예산보다 작다면, 상한액 최소를 올리는것이고,
# 지방세 합이 총예산보다 크면, 상한액 최대를 내리는 것이다.

# 금액
N = int(input()) # 지방의 수
jibang = list(map(int, input().split()))  # 각 지방의 예산 요청 금액을 입력받아 리스트로 저장
# 정수의 입력을 공백기준으로 나누어 리스트에 매핑. map(int)가 분리된 문자열을 정수형으로 변환하는것이다.
budget = int(input()) # 총 예산

# 중간을 하기위한 최소, 최대값
left = 0
right = max(jibang)
answer = -1

while left <= right:
    middle = (left + right) // 2
    
    sum = 0
    for i in range(N):
        sum += min(middle, jibang[i]) # 중간값과 각 지방의 예산요청 금액을 비교해서 작은 값을 더한다.
        
    if sum <= budget:
        answer = middle
        left = middle + 1
    else: 
        right = middle - 1
            
print(answer)

#4
#120 110 140 150
#485
#127
```
