# 그리디 알고리즘(Greedy Algorithm)

- 선택 여러번 -> 각 선택마다 현재의 최적의 선택을 하는 알고리즘

미래의 선택을 고려하지 않고, 현재의 순간만 선택하는것이 핵심.

시간복잡도가 굉장히 빠르게 문제를 해결할 수 있음.

### 핵심은 정렬, 순서에 따라 결과가 다랄지거나, 부분최적이 전체 최적을 보장하지 않으면, DP(동적계획법)으로 전환


# 11399 ATM
https://www.acmicpc.net/problem/11399

```python
N = int(input())

P = list(map(int, input().split()))

P = sorted(P) # 정렬을 해서 돈을 뽑는게 최선

waiting = [0] * N # 배열 만들기
waiting[0] = P[0]

for i in range(1, N):
    waiting[i] = waiting[i-1] + P[i]
    
print(sum(waiting))
```

# 2847 게임을 만든 동준이
https://www.acmicpc.net/problem/2847

```python
# 이전 단계가 다음 단계보다 점수가 높다면 레벨의 점수를 낮춤. 

# 무엇을 선택하냐 -> 각 레벨의 보상
# 어떤 순서로 선택하냐 -> 높은 레벨부터

N = int(input()) # 레벨 수

L = [0] * N # 레벨 저장

for i in range(N):
    L[i] = int(input())
    
count = 0 # 점수 감소 행위
for i in range(N - 2, -1, -1): # 마지막 범위는 수정할 필요가 없다. 
    if L[i] >= L[i+1]:
        count += L[i] - (L[i + 1] - 1)
        L[i] = L[i+1] -1
        
print(count)
        
    
```
