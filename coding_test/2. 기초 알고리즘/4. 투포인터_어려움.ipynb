# 투포인터(Two pointer)

두 개의 포인터를 사용해서, 하나의 배열을 한 번만(or 거의 한 번만) 훑으면서 문제를 푸는 방법

포인터를 하나를더 추가하는 알고리즘 (main, sub).

메인 포인터를 고정하고, 서브포인터를 움직이면서 계산을는방식.

서브포인터가 다 움직이면 메인포인터가 움직인다.

메인포인터와 서브포인터가 같아지면 종료가 된다.

두개의 포인터 사이의 거리: n-1(초기 상태 배열이 n일때), 포인터가 사이 거리가 0이면 종료

n-1부터 0까지 몇번의 덧셈을 하는지 알면, 시간복잡도 계산 가능.

예제) 정렬된 배열 A 원소 중에 합이 x인 두쌍이 있나?
A: [1, 3, 4, 6, 888, 11, 12, 17, 20] x = 20

모든두쌍 o, x

시간복잡도: 두쌍 + -> O(N^2)



```python
N, M = list(map(int, input().split()))

A = list(map(int, input().split()))

start = 0 # 메인 포인터 시작점
end = 0 # 끝 포인터 시작점
sum = A[0]

count = 0

while True:
    # 현재 구간
    if sum == M:
        count +=1
        
    # 구간을 업데이트
    if sum >= M:
        start +=1
        sum -= A[start -1]
    else: # sum < M
        if end == N-1:
            break
        
        end += 1
        sum += A[end]
    
print(count)
```
